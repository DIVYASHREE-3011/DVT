```python
import matplotlib.pyplot as plt

age = [25,45,52,38,60,48]
glucose = [110,145,160,130,180,155]

plt.plot(age, glucose)
plt.title("Age vs Glucose")
plt.xlabel("Age")
plt.ylabel("Glucose")
plt.show()
```

```python
id = [101,102,103,104,105,106]
glucose = [110,145,160,130,180,155]

plt.bar(id, glucose)
plt.title("Patient Glucose Levels")
plt.xlabel("Patient ID")
plt.ylabel("Glucose")
plt.show()
```

```python
gender = ["Female","Male"]
patients = [3,3]

plt.pie(patients, labels=gender, autopct='%1.1f%%')
plt.title("Gender Distribution")
plt.show()
```

```python
glucose = [110,145,160,130,180,155]

plt.hist(glucose, bins=4)
plt.title("Glucose Distribution")
plt.xlabel("Glucose")
plt.ylabel("Number of Patients")
plt.show()
```

```python
glucose = [110,145,160,130,180,155]
hba1c = [5.4,6.5,7.2,6.0,8.0,7.0]

plt.scatter(glucose, hba1c)
plt.title("Glucose vs HbA1c")
plt.xlabel("Glucose")
plt.ylabel("HbA1c")
plt.show()
```

```python
bmi = [22.5,27.8,30.2,25.4,32.1,29.5]

plt.boxplot(bmi)
plt.title("BMI Distribution")
plt.ylabel("BMI")
plt.show()
```

```python
age = [25,45,52,38,60,48]
glucose = [110,145,160,130,180,155]
hba1c = [5.4,6.5,7.2,6.0,8.0,7.0]

plt.plot(age, glucose, label="Glucose")
plt.plot(age, hba1c, label="HbA1c")

plt.xlabel("Age")
plt.ylabel("Values")
plt.title("Glucose and HbA1c")
plt.legend()
plt.show()
```
