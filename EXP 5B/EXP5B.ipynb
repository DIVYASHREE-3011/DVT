```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("C:/Users/svdiv/Downloads/Dataset of Diabetes .csv")

plt.plot(df.index, df["AGE"])

plt.title("Patient Age Trend")
plt.xlabel("Patient Records")
plt.ylabel("Age")

plt.show()
```

```python
gender_count = df["Gender"].value_counts()

plt.bar(gender_count.index, gender_count.values)

plt.title("Gender Distribution")
plt.xlabel("Gender")
plt.ylabel("Number of Patients")

plt.show()
```

```python
class_count = df["CLASS"].value_counts()

plt.pie(
    class_count.values,
    labels=class_count.index,
    autopct='%1.1f%%'
)

plt.title("Patient CLASS Distribution")

plt.show()
```

```python
plt.hist(df["HbA1c"], bins=10)

plt.title("HbA1c Distribution")
plt.xlabel("HbA1c")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.scatter(df["AGE"], df["BMI"])

plt.title("Age vs BMI")
plt.xlabel("Age")
plt.ylabel("BMI")

plt.show()
```

```python
plt.boxplot(df["BMI"].dropna())

plt.title("BMI Box Plot")
plt.ylabel("BMI")

plt.show()
```

```python
plt.plot(df.index, df["Chol"], label="Cholesterol")
plt.plot(df.index, df["TG"], label="Triglycerides")

plt.title("Cholesterol and Triglycerides")
plt.xlabel("Patient Records")
plt.ylabel("Value")

plt.legend()

plt.show()
```
