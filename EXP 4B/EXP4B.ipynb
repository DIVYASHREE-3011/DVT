```python
import pandas as pd

df=pd.read_csv("C:/Users/svdiv/Downloads/Dataset of Diabetes .csv")
print(df.head())
```

```python
print(df.shape)
```

```python
print(df.dtypes)
```

```python
age_data = df[df["AGE"] > 50]

print(age_data)
```

```python
high_hba1c = df[df["HbA1c"] > 7]

print(high_hba1c)
```

```python
sorted_df = df.sort_values(
    by="HbA1c",
    ascending=False
)

print(sorted_df)
```

```python
import matplotlib.pyplot as plt

plt.hist(df["HbA1c"], bins=10)

plt.title("HbA1c Distribution")
plt.xlabel("HbA1c")
plt.ylabel("Frequency")

plt.show()
```

```python
print("Mean =", df["BMI"].mean())
print("Median =", df["BMI"].median())
print("Mode =", df["BMI"].mode()[0])
print("Standard Deviation =", df["BMI"].std())
```

```python
print(df.describe())
```

```python
gender_avg = df.groupby("Gender")["AGE"].mean()

print(gender_avg)
```

```python
gender_avg.plot(kind="bar")

plt.title("Average Age by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Age")

plt.show()
```

```python
corr = df[
    ["AGE", "Urea", "Cr", "HbA1c",
     "Chol", "TG", "HDL", "LDL",
     "VLDL", "BMI"]
].corr()

print(corr)
```

```python
plt.figure(figsize=(10, 6))

plt.imshow(corr, cmap="coolwarm")
plt.colorbar()

plt.xticks(range(len(corr.columns)), corr.columns, rotation=45)
plt.yticks(range(len(corr.columns)), corr.columns)

plt.title("Correlation Matrix")

plt.show()
```
