```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "ID":[101,102,103,104,105,106],
    "Age":[25,45,52,38,60,48],
    "Gender":["Female","Male","Female","Male","Female","Male"],
    "Glucose":[110,145,160,130,180,155],
    "BMI":[22.5,27.8,30.2,25.4,32.1,29.5],
    "HbA1c":[5.4,6.5,7.2,6.0,8.0,7.0]
}

df = pd.DataFrame(data)

print(df)
```

```python
print(df.shape)
```

```python
print(df.dtypes)
```

```python
high_glucose = df[df["Glucose"] > 150]

print(high_glucose)
```

```python
high_hba1c = df[df["HbA1c"] > 6.5]

print(high_hba1c)
```

```python
sorted_df = df.sort_values(
    by="Glucose",
    ascending=False
)

print(sorted_df[["ID","Glucose"]])
```

```python
plt.bar(sorted_df["ID"], sorted_df["Glucose"])

plt.title("Glucose Level Ranking")
plt.xlabel("Patient ID")
plt.ylabel("Glucose")
plt.show()
```

```python
print("Mean =", df["Glucose"].mean())
print("Median =", df["Glucose"].median())
print("Mode =", df["Glucose"].mode()[0])
print("Standard Deviation =", df["Glucose"].std())
```

```python
print(df.describe())
```

```python
gender_avg = df.groupby(
    "Gender"
)["Glucose"].mean()

print(gender_avg)
```

```python
gender_avg.plot(kind="bar")

plt.title("Gender-wise Average Glucose")
plt.xlabel("Gender")
plt.ylabel("Average Glucose")
plt.show()
```

```python
corr = df[
    ["Age","Glucose","BMI","HbA1c"]
].corr()

print(corr)
```
