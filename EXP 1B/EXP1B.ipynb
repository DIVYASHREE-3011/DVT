```python
import pandas as pd

df = pd.read_csv("C:/Users/svdiv/Downloads/Dataset of Diabetes .csv")

data = df[["AGE", "BMI", "HbA1c", "CLASS"]]

print(data.head())
```

```python
print(data.isnull().sum())
print("Duplicates =", data.duplicated().sum())

data = data.drop_duplicates()
```

```python
data["CLASS"] = data["CLASS"].astype(str)

print(data.dtypes)
```

```python
print(data.groupby("CLASS")["AGE"].mean())
print(data.groupby("CLASS")["BMI"].mean())
print(data.groupby("CLASS")["HbA1c"].mean())
```

```python
import matplotlib.pyplot as plt

plt.scatter(data["HbA1c"], data["BMI"])

plt.title("HbA1c vs BMI")
plt.xlabel("HbA1c")
plt.ylabel("BMI")
plt.show()
```

```python
print("Diabetes Class Distribution")
print(data["CLASS"].value_counts())

print("\nAverage HbA1c by Class")
print(data.groupby("CLASS")["HbA1c"].mean())
```
